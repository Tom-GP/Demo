Getting the group of data base on temperature ranges

```python
def group_temp(df, idx, col):
  if 80 <= df.loc[idx, col] <= 90:
    return '80-90'
  elif 50 <= df.loc[idx, col] <= 60:
    return '50-60'
  else:
    return 'others'
```

```python
g = df3.groupby(lambda idx: group_temp(df3, idx, 'temperature'))
for key, data in g:
  print(f'\nTemp b/w {key}:\n')
  print(data)
```
