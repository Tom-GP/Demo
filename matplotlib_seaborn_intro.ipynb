```python
import pandas as pd
df = pd.read_excel('linechart.xlsx')
df.head()
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(12, 7))
plt.plot(df.Quarter, df.Fridge, color='red', label='Fridge')
plt.plot(df['Quarter'], df['Dishwasher'], label='Dishwasher')
plt.title('PRODUCT SALES')
plt.xlabel('Quarter')
plt.ylabel('Sales')
plt.legend()
plt.show()
```

```python
total_sales = df[['Dishwasher', 'Fridge', 'Washing Machine']].sum()
total_sales
```

```python
plt.pie(total_sales, labels=total_sales.index, autopct='%1.1f%%', explode=(0, 0.1, 0),
        shadow=True, startangle=160)
plt.show()
```

```python
df.plot(kind='bar',  x='Quarter')
plt.xticks(rotation=45)
plt.show()
```
